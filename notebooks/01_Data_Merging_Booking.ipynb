##Booking

```python
booking_path = data_dir /'bookings'
```

###Legacy

```python
def read_city_csv_legacy(city_path):
    df_list = []

    for item in city_path.iterdir():
        item_path = city_path /item.name 
        #print(item_path)
        df = pd.read_csv(item_path, compression="gzip", encoding="latin-1")
        df_list.append(df)

    return pd.concat(df_list, ignore_index=True)
```

####Ahmedabad

```python
ahmd_path = booking_path / 'legacy' / 'ahmedabad'
df_ahmed = read_city_csv_legacy(ahmd_path)
df_ahmed.head()
```

```python
def read_json_nova(city_path):
    df_list = []

    for item in city_path.iterdir():
        item_path = city_path /item.name 
        df = pd.read_json(item_path, lines=True, compression="gzip")
        patient_df = pd.json_normalize(df['patient'])
        df = pd.concat( [df.drop(columns=['patient']), patient_df], axis=1 )
        df = df.explode('tests', ignore_index=True)
        tests_df = pd.json_normalize(df['tests'])
        tests_df = tests_df.add_prefix('test_')
        df = pd.concat([df.drop(columns=['tests']), tests_df], axis=1)
        df_list.append(df)

    return pd.concat(df_list, ignore_index=True)
```

####Chennai

```python
chen_nova_path = booking_path /'nova' /'chennai'

df_chen_nova = read_json_nova(chen_nova_path)
df_chen_nova
```

####Jaipur

```python
jai_nova_path = booking_path /'nova' /'jaipur'

df_jai_nova = read_json_nova(jai_nova_path)
df_jai_nova
```

##Merged CSV for Legacy and Nova

###Legacy

```python
legacy_path = booking_path / 'legacy'
```

```python
legacy_df_list = [df_ahmed,df_beng,df_chen,df_hyd,df_jai,df_pun]
df_ahmed.to_csv(legacy_path / "ahmedabad.csv", index=False)
df_beng.to_csv(legacy_path / "bengaluru.csv", index=False)
df_chen.to_csv(legacy_path / "chennai.csv", index=False)
df_hyd.to_csv(legacy_path / "hyderabad.csv", index=False)
df_jai.to_csv(legacy_path / "jaipur.csv", index=False)
df_pun.to_csv(legacy_path / "pune.csv", index=False)
df_legacy_merged = pd.concat(legacy_df_list, ignore_index=True)
df_legacy_merged.to_csv(legacy_path / "merged.csv", index=False)
```

###Nova

```python
nova_path = booking_path / 'nova'
```

```python
nova_df_list = [df_chen_nova,df_jai_nova]
df_chen_nova.to_csv(nova_path / "chennai.csv", index=False)
df_jai_nova.to_csv(nova_path / "jaipur.csv", index=False)
df_nova_merged = pd.concat(nova_df_list, ignore_index=True)
df_nova_merged.to_csv(nova_path / "merged.csv", index=False)
```
